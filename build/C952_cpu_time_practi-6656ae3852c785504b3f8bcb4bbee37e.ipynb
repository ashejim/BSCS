**Handy formulas** (IC = instruction count, CPI = cycles per instruction, $f$ = clock rate):
- CPU time $= \dfrac{IC \times CPI}{f} = IC \times CPI \times t_{cycle} = \dfrac{\text{clock cycles}}{f}$
- Weighted CPI $= \sum (\text{fraction}_i \times \text{cycles}_i)$
- Speedup $= \dfrac{\text{old time}}{\text{new time}}$; a speedup of $s$ is a $(s-1)\times 100\%$ improvement
- Note: `7000e9` means $7000 \times 10^{9}$.
