**Handy formulas** ($k$ = stages, $t$ = cycles per stage, $n$ = number of instruction sets):
- Total pipelined cycles $= (k + n - 1)\,t$
- Average cycles per set $= \dfrac{(k + n - 1)\,t}{n}$
- Speedup vs. non-pipelined $= \dfrac{n\,k}{k + n - 1}$, which approaches $k$ as $n$ grows large
- Laundry-style: (sum of all stage times) $+$ (loads $-\,1$) $\times$ (slowest stage)
